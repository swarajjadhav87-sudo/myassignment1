1. Create a Git working directory with feature1.txt and feature2.txt in the master branch
2. Create 3 branches develop, feature1 and feature2
3. In develop branch create develop.txt, do not stage or commit it
4. Stash this file and check out to feature1 branch
5. Create new.txt file in feature1 branch, stage and commit this file
6. Checkout to develop, unstash this file and commit
7. Please submit all the Git commands used to do the above step>

```python
# Step 1: Create repo folder
mkdir task2
cd task2
git init

# Step 2: Create initial files and commit
touch feature1.txt feature2.txt
git add feature1.txt feature2.txt
git commit -m "Initial commit with feature1.txt and feature2.txt"

# Step 3: Create branches
git branch develop
git branch feature1
git branch feature2

# Step 4: Work in develop branch
git checkout develop
touch develop.txt          # create file
git stash                  # stash uncommitted file

# Step 5: Work in feature1 branch
git checkout feature1
touch new.txt
git add new.txt
git commit -m "Added new.txt in feature1 branch"

# Step 6: Back to develop branch, unstash and commit
git checkout develop
git stash pop
git add develop.txt
git commit -m "Added develop.txt in develop branch"

# Step 7: Push all branches to GitHub
git remote add origin https://github.com/<your-username>/task2-repo.git
git branch -M main          # if default branch is main
git push -u origin main
git push -u origin develop
git push -u origin feature1
git push -u origin feature2
```

3. Put develop.txt in develop branch, f1.txt and f2.txt in f1 and f2 respectively
4. Push all these branches to GitHub

5. On local delete f2 branch
6. Delete the same branch on GitHub as well3

```python
mkdir task4
cd task4
git init
echo "This is master branch file" > main.txt
git add main.txt
git commit -m "Initial commit: added main.txt"

git branch develop
git branch f1
git branch f2

git checkout develop
echo "This is develop branch file" > develop.txt
git add develop.txt
git commit -m "Added develop.txt in develop branch"

git checkout f1
echo "This is f1 branch file" > f1.txt
git add f1.txt
git commit -m "Added f1.txt in f1 branch"

git checkout f2
echo "This is f2 branch file" > f2.txt
git add f2.txt
git commit -m "Added f2.txt in f2 branch"

git remote add origin https://github.com/<your-username>/task4-repo.git
git push -u origin master
git push -u origin develop
git push -u origin f1
git push -u origin f2

# Delete f2 branch
git branch -d f2
git push origin --delete f2
```
