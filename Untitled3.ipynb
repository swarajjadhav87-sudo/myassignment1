3) Compare and contrast mutable and immutable objects in Python with examples0
Mutable Objects

Definition:
Mutable objects can be modified in place after they are created.

Characteristics

Values can be changed

Same memory location is used

Changes affect all references to the object

Common Mutable Objects

list

dict

set

bytearray

a = [1, 2, 3]
b = a          # both refer to same object

a.append(4)
print(a)       # [1, 2, 3, 4]
print(b)       # [1, 2, 3, 4]

Immutable Objects

Definition:
Immutable objects cannot be changed after creation.
Any modification creates a new object.

Characteristics

Values cannot be altered

New memory location is created on modification

Safe from unintended changes

Common Immutable Objects

int

float

str

tuple

frozenset

x = 10
y = x

x = x + 5
print(x)       # 15
print(y)       # 10

Comparison between Mutable and Immutable Objects in Python (Point-wise)

Changeability
Mutable objects can be modified after they are created, whereas immutable objects cannot be changed once created.

Memory Behavior
Mutable objects retain the same memory location when modified, while immutable objects create a new memory location for every modification.

Effect on References
Changes in a mutable object affect all references pointing to it, but changes in an immutable object do not affect other references.

Performance
Mutable objects are more efficient when frequent updates are required, whereas immutable objects may consume more memory when modified repeatedly.

Safety
Mutable objects are more prone to accidental changes, while immutable objects are safer and more predictable.                             |
Hashability
Mutable objects are generally not hashable and cannot be used as dictionary keys, whereas immutable objects are hashable.

Examples
Mutable objects include lists, dictionaries, and sets.
Immutable objects include integers, strings, tuples, and floats.


Type Casting in Python

Type casting in Python is the process of converting one data type into another.
It is mainly used when we need to perform operations on different data types or take user input in a required format.
Types of Type Casting in Python
1. Implicit Type Casting

Python automatically converts one data type into another without user intervention.

Example
a = 10        # int
b = 2.5       # float
c = a + b     # int converted to float automatically
print(c)      # 12.5
print(type(c))  # <class 'float'>
2. Explicit Type Casting

The programmer manually converts one data type into another using built-in functions.

Common Type Casting Functions:

int()

float()

str()

list()

tuple()

set()

# String to Integer
x = "25"
y = int(x)
print(y, type(y))

# Integer to Float
a = 10
b = float(a)
print(b, type(b))

# Integer to String
num = 100
s = str(num)
print(s, type(s))


6) How do conditional statements work in Python? Illustrate with examples
Conditional Statements in Python

Conditional statements in Python are used to make decisions and execute different blocks of code based on whether a condition is True or False.

Python uses the keywords if, elif, and else for decision making.
How Conditional Statements Work

The if condition is checked first.

If it is True, the corresponding block of code executes.

If the if condition is False, Python checks the elif condition(s) (if present).

If none of the conditions are True, the else block executes.

Python uses indentation to define blocks of code

Example 1: Simple if Statement
age = 20

if age >= 18:
    print("Eligible to vote")

    Example 2: if–else Statement
num = 5

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")

    Example 3: if–elif–else Ladder
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
    

7 Describe the different types of loops in Python and their use cases with examples.
Loops in Python

Loops in Python are used to execute a block of code repeatedly until a certain condition is met.

Python mainly provides two types of loops:

for loop

while loop

Python also supports nested loops and loop control statements.
1. for Loop

Description:
The for loop is used to iterate over a sequence such as a list, tuple, string, or range.

Use cases:

Traversing elements of a collection

Repeating tasks a fixed number of times
for i in range(5):
    print(i)
2. while Loop

Description:
The while loop executes as long as the given condition is True.

Use cases:

When the number of iterations is not known beforehand

Repeating tasks until a condition changes

Example:
count = 1

while count <= 5:
    print(count)
    count += 1
3. Nested Loops

Description:
A loop inside another loop is called a nested loop.

Use cases:

Working with matrices

Pattern printing

Example:

for i in range(3):
    for j in range(2):
        print(i, j)
